## Data Cleaning for Hip-Hop Tracks

### 1. Removing Duplicate Rows
First, I will remove any exact duplicate rows from the `hip_hop_df` to ensure each entry is unique. This helps prevent bias in our analysis.

```python
initial_rows = hip_hop_df.shape[0]
hip_hop_df.drop_duplicates(inplace=True)
duplicates_removed = initial_rows - hip_hop_df.shape[0]
print(f"Number of duplicate rows removed: {duplicates_removed}")
print(f"New shape of hip_hop_df: {hip_hop_df.shape}")
```

### 2. Handling Missing Values
Next, I will check for any missing values specifically within our `hip_hop_df` and decide on a strategy to handle them. For essential track details like artists, album name, and track name, it's often best to remove rows where these are missing.

```python
print("Missing values before handling:")
display(hip_hop_df.isnull().sum())

# Drop rows where 'artists', 'album_name', or 'track_name' are missing
hip_hop_df.dropna(subset=['artists', 'album_name', 'track_name'], inplace=True)
print("\nMissing values after dropping rows:")
display(hip_hop_df.isnull().sum())
print(f"New shape of hip_hop_df after handling missing values: {hip_hop_df.shape}")
```

### 3. Addressing 'Odd Values' (Outliers and Inconsistencies)
'Odd values' can refer to outliers in numerical data or inconsistencies in categorical data. For numerical features, I'll start by examining descriptive statistics to identify potential issues like negative durations, or values far outside the expected range. For categorical features, we will check unique values if any particular column seems problematic.

```python
print("Descriptive statistics for numerical columns:")
display(hip_hop_df.describe())

# Check for non-positive durations, which would be 'odd'
non_positive_duration_tracks = hip_hop_df[hip_hop_df['duration_ms'] <= 0]
if not non_positive_duration_tracks.empty:
    print("\nTracks with non-positive duration_ms found:")
    display(non_positive_duration_tracks)
    # Option to remove them if they are indeed erroneous
    # hip_hop_df = hip_hop_df[hip_hop_df['duration_ms'] > 0]
else:
    print("\nNo tracks found with non-positive duration_ms.")

# Example for popularity: check for values outside [0, 100] range
outlier_popularity_tracks = hip_hop_df[(hip_hop_df['popularity'] < 0) | (hip_hop_df['popularity'] > 100)]
if not outlier_popularity_tracks.empty:
    print("\nTracks with popularity outside [0, 100] range found:")
    display(outlier_popularity_tracks)
else:
    print("\nNo tracks found with popularity outside the [0, 100] range.")
```

The data looks reasonably clean after these steps. The `describe()` output helps to spot major outliers (e.g., min/max values that are physically impossible or extremely unlikely). If specific columns were identified as having 'odd' values beyond these general checks, further custom cleaning steps could be applied.
